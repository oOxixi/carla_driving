# Read-only reproduction evidence viewer
This notebook performs no install, download, model load, or inference.


```python
from pathlib import Path
import hashlib, json, os
RUN_DIR = Path(os.environ.get("CARLA_REPRO_RUN_DIR", "../metrics/reference_5070")).resolve()
RUN_DIR
```

```python
def load_json(relative):
    return json.loads((RUN_DIR / relative).read_text(encoding="utf-8"))
def sha256(relative):
    return hashlib.sha256((RUN_DIR / relative).read_bytes()).hexdigest()
manifest = load_json("run_manifest.json")
environment = load_json("environment.json")
model = load_json("model_manifest.json")
for relative, expected in manifest.get("files", {}).items():
    assert sha256(relative) == expected, relative
manifest, environment, model
```

```python
latency = load_json("metrics/latency.json")
latency_table = {key: latency.get(key) for key in ("count", "mean_ms", "p50_ms", "p95_ms", "max_ms", "status", "scope")}
latency_table
```
